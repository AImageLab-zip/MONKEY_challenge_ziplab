```python
import tarfile
import os

def convert_pth_to_targz(pth_file_path, output_file_path):
    # Step 1: Check if .pth file exists
    if not os.path.exists(pth_file_path):
        print(f"Error: {pth_file_path} does not exist.")
        return
    
    # Step 2: Create a tar.gz file
    with tarfile.open(output_file_path, "w:gz") as tar:
        # Add the .pth file to the archive
        tar.add(pth_file_path, arcname=os.path.basename(pth_file_path))
    
    print(f"Model converted and saved to {output_file_path}")

# Example usage
pth_file_path = './outputs/model_final.pth'  # Path to your .pth model
output_file_path = './outputs/model_final.tar.gz'  # Desired output tar.gz file

convert_pth_to_targz(pth_file_path, output_file_path)
```
